<div style="background-color:#69CFCF; color: black; border-left: 6px solid #006064; padding: 15px; border-radius: 5px;">
    <h3 style="color:#006064; margin-top:0;">📝 Planteamiento del Problema</h3>
    <p><strong>Contexto:</strong> Descripción del problema.</p>
    <p><strong>Ecuación Diferencial Ordinaria (EDO):</strong></p>
    $$ \frac{dP}{dt} = rP \left(1 - \frac{P}{K}\right) $$
    <p>Donde:</p>
    <ul>
        <li>$P(t)$: Población / Concentración en el tiempo $t$.</li>
        <li>$r$: Tasa de crecimiento / decaimiento.</li>
        <li>$K$: Capacidad de carga del sistema.</li>
    </ul>
</div>

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from scipy.integrate import odeint
import seaborn as sns

print("Librerías cargadas")

# Definición de variables simbólicas
t = sp.symbols('t')
P = sp.Function('P')(t)
r, K, P0 = sp.symbols('r K P0')

# 1. Definir la Ecuación Diferencial
# Ejemplo: Modelo Logístico -> dP/dt - rP(1 - P/K) = 0
edo = sp.Eq(P.diff(t), r * P * (1 - P/K))

# 2. Resolver la EDO simbólicamente
sol_general = sp.dsolve(edo, P)

# 3. Aplicar condiciones iniciales (si existen)
# P(0) = P0
ics = {P.subs(t, 0): P0}
constantes = sp.solve(sol_general.subs(t, 0).subs(P.subs(t, 0), P0), 'C1')
# Nota: La resolución de C1 depende de la forma de la solución, ajustar según necesidad.

print("--- Solución General Simbólica ---")
display(sol_general)
```

```python
# --- PARÁMETROS DEL MODELO (Ajustar según ejercicio) ---
r_val = 0.5    # Tasa de crecimiento (ej. 1/año)
K_val = 1000   # Capacidad de carga (ej. toneladas de biomasa)
P0_val = 100   # Población inicial
t_max = 20     # Tiempo máximo de simulación

# Definición de la función para odeint
def modelo(y, t, r, k):
    P = y
    dydt = r * P * (1 - P/k) # Ecuación logística
    return dydt

# Vector de tiempo
t_num = np.linspace(0, t_max, 100)

# Resolución numérica
solucion = odeint(modelo, P0_val, t_num, args=(r_val, K_val))
P_num = solucion.flatten()

print(f"Solución numérica calculada para t=0 a t={t_max}.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.integrate import odeint
from IPython.display import HTML

# 1. Configuración del Modelo
def modelo_logistico(P, t, r, K):
    return r * P * (1 - P/K)

# Parámetros fijos
K = 1000
P0 = 50
t_max = 30
r = 0.5

# Vector de tiempo y solución "base"
t = np.linspace(0, t_max, 200)
P_final = odeint(modelo_logistico, P0, t, args=(r, K)).flatten()

# 2. Configurar el escenario vacío
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, t_max)
ax.set_ylim(0, K * 1.2)
ax.set_xlabel('Tiempo (t)', fontsize=12)
ax.set_ylabel('Población', fontsize=12)
ax.set_title(f'Simulación Dinámica: Crecimiento Logístico\n$r={r}, K={K}$', fontsize=14, color='#003366', fontweight='bold')
ax.grid(True, linestyle=':', alpha=0.6)

# Línea de Capacidad de Carga (Fija)
ax.axhline(y=K, color='#00a86b', linestyle='--', label='Capacidad de Carga ($K$)')

# Elementos que se moverán
linea, = ax.plot([], [], lw=3, color='#0077be', label='Población')
punto_actual, = ax.plot([], [], 'o', color='#d9534f', markersize=8) # Un punto rojo que sigue la curva
texto_valor = ax.text(0.02, 0.95, '', transform=ax.transAxes, fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

ax.legend(loc='lower right')

# 3. Función de Animación (Frame a Frame)
def animar(i):
    # i va de 0 a len(t)
    x_data = t[:i]
    y_data = P_final[:i]
    
    linea.set_data(x_data, y_data)
    
    # Si hay datos, actualizamos el punto rojo en la cabeza
    if i > 0:
        punto_actual.set_data([t[i-1]], [P_final[i-1]]) # Deben ser listas/arrays
        texto_valor.set_text(f'Tiempo: {t[i-1]:.1f}\nPoblación: {P_final[i-1]:.0f}')
    
    return linea, punto_actual, texto_valor

# 4. Generar la animación
print("🎥 Generando animación, espera unos segundos...")
anim = FuncAnimation(fig, animar, frames=len(t), interval=50, blit=True)

# Cerrar la figura estática para que no salga duplicada
plt.close()

# 5. Mostrar como video HTML interactivo (Play/Pause)
HTML(anim.to_jshtml())
```
